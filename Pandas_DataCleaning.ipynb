
Renaming columns


```python
begin_df = begin_df.rename(columns={
    "onHand": "BeginOnHand",
    "Price": "BeginPrice"
})

end_df = end_df.rename(columns={
    "onHand": "EndOnHand",
    "Price": "EndPrice"
})

print(begin_df.columns)
print(end_df.columns)


```

Merging


```python
begin_end_merged=  pd.merge(begin_df,end_df, on="InventoryId" , how="outer")
print(begin_end_merged.head(),"\n")
begin_end_merged.shape
```

Filling NaN values fo BeginOnhand and EndOnHand  , BeginPrice and EndPrice


```python
begin_end_merged["BeginOnHand"] = begin_end_merged["BeginOnHand"].fillna(0)
begin_end_merged["EndOnHand"] = begin_end_merged["EndOnHand"].fillna(0)
begin_end_merged["BeginPrice"] = begin_end_merged["BeginPrice"].fillna(0)
begin_end_merged["EndPrice"] = begin_end_merged["EndPrice"].fillna(0)


print(begin_end_merged.head())
```

Changing Date datatype from str to date


```python
begin_end_merged[["startDate", "endDate"]] = begin_end_merged[["startDate", "endDate"]].apply(pd.to_datetime)
print(begin_end_merged.head())
```

Combining appropriate columns together for data validation

```python
begin_end_merged["Store"] = begin_end_merged["Store_x"].combine_first(begin_end_merged["Store_y"])

begin_end_merged["City"] = begin_end_merged["City_x"].combine_first(begin_end_merged["City_y"])

begin_end_merged["Brand"] = begin_end_merged["Brand_x"].combine_first(begin_end_merged["Brand_y"])

begin_end_merged["Description"] = begin_end_merged["Description_x"].combine_first(begin_end_merged["Description_y"])

begin_end_merged["Size"] = begin_end_merged["Size_x"].combine_first(begin_end_merged["Size_y"])
print(begin_end_merged.head())
```

Dropping columns that arent needed

```python

begin_end_merged = begin_end_merged.drop([
    "Store_x", "Store_y",
    "City_x", "City_y",
    "Brand_x", "Brand_y",
    "Description_x", "Description_y",
    "Size_x", "Size_y"
], axis=1)

```

```python
begin_end_price_df = pd.merge(
    begin_end_merged,
    price_df,
    on="Brand",
    how="left",
    suffixes=("_inventory", "_purchase")
)


print(begin_end_price_df.head())
```

```python
"""begin_end_price_df = begin_end_price_df.drop([
    "Description_purchase",
    "Size_purchase"
], axis=1)

begin_end_price_df = begin_end_price_df.rename(columns={
    "Description_inventory": "Description",
    "Size_inventory": "Size"
})"""
```

```python
begin_end_price_df["StockChange"] = begin_end_price_df["EndOnHand"] - begin_end_price_df["BeginOnHand"]

begin_end_price_df["BeginInventoryValue"] = begin_end_price_df["BeginOnHand"] * begin_end_price_df["BeginPrice"]

begin_end_price_df["EndInventoryValue"] = begin_end_price_df["EndOnHand"] * begin_end_price_df["EndPrice"]

begin_end_price_df["Profit"] = begin_end_price_df["EndPrice"] - begin_end_price_df["PurchasePrice"]

begin_end_price_df["ProfitMargin"] = ((begin_end_price_df["EndPrice"] - begin_end_price_df["PurchasePrice"]) / begin_end_price_df["PurchasePrice"]) * 100
```

Group By Analysis


```python
print(begin_end_price_df.groupby("Brand")["EndInventoryValue"].sum().sort_values(ascending=False))
```

```python
print(begin_end_price_df.groupby("City")["EndInventoryValue"].sum().sort_values(ascending=False))
```
